## 파생변수 생성

### 흡연 여부 컬럼 생성

```python
# 흡연 여부 컬럼 생성 
df_old['흡연 여부'] = np.nan

# 흡연 경험 1: 예 2: 아니오 일반담배 흡연 경험 1: 5갑 미만 2: 5갑 이상 3: 피운적x

# 흡연 경험이 1이면서 일반 담배 흡연 경험이 2인 값
# 일반담배 현재 흡연 여부가 1 또는 2인 값
cond1 = (df_old['흡연 경험'] == 1 & (df_old['일반담배 흡연 경험'] == 2) & (df_old['일반담배 현재 흡연 여부'].isin([1, 2]))) 

# 흡연 경험 1, 일반 담배 경험 2, 일반담배 현재 흡연 여부가 3인 경우 값
cond2 = (df_old['흡연 경험'] == 1 & (df_old['일반담배 흡연 경험'] == 2) & (df_old['일반담배 현재 흡연 여부'] == 3))

# 흡연경험이 1이지만 힐반담배 흡연 경험이 1또는 3인 경우 값
cond3 = (df_old['흡연 경험'] == 1) & (df_old['일반담배 흡연 경험'].isin([1, 3]))

# 흡연경험 2인 경우
cond4 = (df_old['흡연 경험'] == 2)

# 궐련형 전자담배 흡연 경험 1: 예 2 : 아니오 , 궐련형 전자담배 현재 흡연 여부 : 1 매일핌 2: 가끔핌 3: 현재안핌

# 흡연 경험이 1이면서 궐련형 전담경험 1 또는 2 값, 현재 궐련형 담배 흡연 1 또는 2 인 경우 
cond5 = (df_old['흡연 경험'] == 1) & (df_old['궐련형 전자담배 흡연 경험'].isin([1, 2])) & (df_old['궐련형 전자담배 현재 흡연 여부'].isin([1, 2]))

# 흡연 경험이 1 이면서 궐련형 전담경험 1이지만 궐련형 전자담배 현재 흡연 여부가 3인 경우
cond6 = (df_old['흡연 경험'] == 1) & (df_old['궐련형 전자담배 흡연 경험'].isin([1, 2])) & (df_old['궐련형 전자담배 현재 흡연 여부'] == 3)

# 일반담배 기준 흡연 여부 설정
df_old.loc[cond1, '흡연 여부'] = 1  # 현재 일반담배 흡연 중이면 1
df_old.loc[cond2, '흡연 여부'] = 0  # 과거 흡연자지만 현재는 비흡연이면 0
df_old.loc[cond3, '흡연 여부'] = 0  # 일반담배 흡연 경험 없음/기타 → 비흡연
df_old.loc[cond4, '흡연 여부'] = 0  # 아예 흡연 경험 없음 → 비흡연

# 궐련형 전자담배 기준 흡연 여부 설정
df_old.loc[cond5, '흡연 여부'] = 1  # 현재 궐련형 전자담배 흡연 중이면 1
df_old.loc[cond6, '흡연 여부'] = 0  # 과거 흡연자이지만 현재는 비흡연이면 0
```

### 일 평균 흡연량

```python
# 계산할 때만 사용으로 카피해두기 
copy_01z = df_old['일반담배 하루 평균 흡연량'].copy()
copy_11z = df_old['궐련형 전자담배 하루 평균 흡연량'].copy()

# 계산할 때 이상치 방지로 777 888 999 값들 -1로 변경하기
df_old['일반담배 하루 평균 흡연량'] = df_old['일반담배 하루 평균 흡연량'].replace([777, 888, 999], -1)
df_old['궐련형 전자담배 하루 평균 흡연량'] = df_old['궐련형 전자담배 하루 평균 흡연량'].replace([777, 888, 999], -1)

# 흡연자만 계산 하고 비흡연자는 0값
df_old['일 평균 흡연량'] = 0  
mask_smoker = df_old['흡연 여부'] == 1

# 흡연량 전부 더하기
# 지피티니가 추천해준 clip(lower0)은 음수를 0으로 바꿔서 덧셈하는 기능이라함 
df_old.loc[mask_smoker, '일 평균 흡연량'] = (
    df_old.loc[mask_smoker, '일반담배 하루 평균 흡연량'].clip(lower=0) +
    df_old.loc[mask_smoker, '궐련형 전자담배 하루 평균 흡연량'].clip(lower=0)
)

# 다시 원래 값으로 변경하기
df_old['일반담배 하루 평균 흡연량'] = copy_01z
df_old['궐련형 전자담배 하루 평균 흡연량'] = copy_11z
```

### 성별 평균 흡연비율

```python
# 흡연 여부가 0 또는 1인 행만 필터링 (이상치 -1 등 제거)
df_valid = df_old[df_old['흡연 여부'].isin([0, 1])]

# 성별 기준 흡연 여부 평균 계산 (남성=1, 여성=2)
흡연율_남 = df_valid.loc[df_valid['성별'] == 1, '흡연 여부'].mean()
흡연율_여 = df_valid.loc[df_valid['성별'] == 2, '흡연 여부'].mean()

# 성별에 따라 흡연율 값을 새 컬럼에 입력
df_old['성별 흡연비율'] = np.where(df_old['성별'] == 1, 흡연율_남, np.nan)
df_old['여성 흡연비율'] = np.where(df_old['성별'] == 2, 흡연율_여, np.nan)

# 비율로 변환 소수점 둘째자리까지 반올림
df_old['성별 흡연비율'] = (df_old['성별 흡연비율'] * 100).round(2)
df_old['여성 흡연비율'] = (df_old['여성 흡연비율'] * 100).round(2)

# 여성인 경우 '성별 흡연비율' 값을 '여성 흡연비율' 값으로 덮어씀
df_old.loc[df_old['성별'] == 2, '성별 흡연비율'] = df_old.loc[df_old['성별'] == 2, '여성 흡연비율']

# 임시로 만든 여성 흡연비율 컬럼 제거
df_old.drop(columns=['여성 흡연비율','일반담배 흡연 경험', '일반담배 현재 흡연 여부', '일반담배 하루 평균 흡연량',
                     '궐련형 전자담배 흡연 경험', '궐련형 전자담배 현재 흡연 여부', '궐련형 전자담배 하루 평균 흡연량'], 
                     inplace=True)
```

### 우울증상으로 합치기

```python
depression_col = [
    '직무 흥미 저하',
    '우울감',
    '수면장애',
    '무기력',
    '식이장애'
]

for col in depression_col:
    print(f"\n[{col}]")
    print(df_old[col].value_counts(dropna=False).sort_index())

# 우울증상 지표 5가지를 '우울증상'이라는 하나의 파생변수로 생성
df_old['우울증상'] = df_old[depression_col].sum(axis=1)
df_old = df_old.drop(columns = depression_col)
```

### 수면 컬럼 값 변경 및 제거

```python
# 취침(분)이 50 이상이면 취침(시)에 1을 더하고 24시 까지 표기, 25로 넘어갈 시 1로 변경
df_old['취침(시)'] = (df_old['취침(시)'] + (df_old['취침(분)'] >= 50).astype(int)) % 24

# 기상(분)이 50 이상이면 기상(시)에 1을 더하고 24시 까지 표기, 25로 넘어갈 시 1로 변경
df_old['기상(시)'] = (df_old['기상(시)'] + (df_old['기상(분)'] >= 50).astype(int)) % 24

# 수면소요(시) 값을 분으로 바꾸고 수면요소(분)에 더하기, -1일 때 계산 x
mask1 = (df_old['수면소요(시)'] != -1) & (df_old['수면소요(분)'] != -1)
df_old.loc[mask1, '수면소요(분)'] = df_old.loc[mask1, '수면소요(시)'] * 60 + df_old.loc[mask1, '수면소요(분)']

# 실수면(시) 값을 분으로 바꾸고 실수면(분)에 더하기, -1일 때 계산 x
mask2 = (df_old['실수면(시)'] != -1) & (df_old['실수면(분)'] != -1)
df_old.loc[mask2, '실수면(분)'] = df_old.loc[mask2, '실수면(시)'] * 60 + df_old.loc[mask2, '실수면(분)']

# 고강도 신체활동(시) 값을 분으로 바꾸고 고강도 운동 실천(분)에 더하기, -1일 때 계산 x
mask3 = (df_old['고강도 신체활동(시)'] != -1) & (df_old['고강도 신체활동(분)'] != -1)
df_old.loc[mask3, '고강도 신체활동(분)'] = df_old.loc[mask3, '고강도 신체활동(시)'] * 60 + df_old.loc[mask3, '고강도 신체활동(분)']

# 걷기 실천(시) 값을 분으로 바꾸고 걷기 실천(분)에 더하기, -1일 때 계산 x
mask4 = (df_old['걷기 실천(시)'] != -1) & (df_old['걷기 실천(분)'] != -1)
df_old.loc[mask4, '걷기 실천(분)'] = df_old.loc[mask4, '걷기 실천(시)'] * 60 + df_old.loc[mask4, '걷기 실천(분)']

# 정수형으로 변환
df_old[['취침(시)', '기상(시)', '수면소요(분)', '실수면(분)', '고강도 신체활동(분)', '걷기 실천(분)']] = df_old[['취침(시)', '기상(시)', '수면소요(분)', '실수면(분)', '고강도 신체활동(분)', '걷기 실천(분)']].astype(int)

# 안 쓰는 값들 컬럼 제거
df_old.drop(columns=['취침(분)', '기상(분)', '수면소요(시)', '실수면(시)', '고강도 신체활동(시)', '걷기 실천(시)'] , inplace=True)
```

```python
# 주관적 스트레스 수준 1 : 대단히 많이 느낌, 2 : 많이 느낌, 3 : 조금 느낌, 4 : 거의 안 느낌
# 1이 안 좋은거라 1부터 0.25~ 4까지 1
df_old['스트레스 수준'] = df_old['스트레스 수준'].map({1 : 0.25, 2 : 0.5, 3 : 0.75, 4 : 1, -1 : 0.1})

# 자기비하 1 : 전혀 아님 2 : 여러날  3 : 일주일 이상 4 : 거의 매일
# 1이 좋음 1부터 1~0.25 4까지 -1은 0.1 
df_old['자기비하'] = df_old['자기비하'].map({1 : 1, 2 : 0.75, 3 : 0.5, 4 : 0.25, -1 : 0.1})

# 집중력저하 1 : 전혀 아님 2 : 여러날  3 : 일주일 이상 4 : 거의 매일
# 1이 좋음 1부터 1~0.25 4까지 -1은 0.1 
df_old['집중력저하'] = df_old['집중력저하'].map({1 : 1, 2 : 0.75, 3 : 0.5, 4 : 0.25, -1 : 0.1})

# 활동둔화 1 : 전혀 아님 2 : 여러날  3 : 일주일 이상 4 : 거의 매일
# 1이 좋음 1부터 1~0.25 4까지 -1은 0.1 
df_old['활동둔화'] = df_old['활동둔화'].map({1 : 1, 2 : 0.75, 3 : 0.5, 4 : 0.25, -1 : 0.1})

# 자살충동 1 : 전혀 아님 2 : 여러날  3 : 일주일 이상 4 : 거의 매일
# 1이 좋음 1부터 1~0.25 4까지 -1은 0.1 
df_old['자살충동'] = df_old['자살충동'].map({1 : 1, 2 : 0.75, 3 : 0.5, 4 : 0.25, -1 : 0.1})
```

#### 수면 관련 컬럼

## 비만 BMI 

```python
import matplotlib.pyplot as plt
import seaborn as sns

# BMI 수치별로 정상 비만 구별
def bmi(x):
    if pd.isna(x) or x in [-1, 777, 888, 999]:
        return '삭제'
    elif x < 18.5:
        return '저체중'
    elif 18.5 <= x <= 22.9:
        return '정상'
    elif 23.0 <= x <= 24.9:
        return '과체중'
    elif 25.0 <= x <= 29.9:
        return '비만 1단계'
    elif 30.0 <= x <= 34.9:
        return '비만 2단계'
    else:
        return '고도비만'

df_old['BMI 수치'] = df_old['BMI'].apply(bmi)

# 한글, 마이너스 깨짐 방지~
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 데이터 분리
df_co = df_old[df_old['인지장애 경험 여부'] == 1]
df_nco = df_old[df_old['인지장애 경험 여부'] == 2]

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5), sharey=True)

# 인지장애 경험자
plot1 = sns.countplot(
    x=df_co['BMI 수치'],
    order=['저체중', '정상', '과체중', '비만 1단계', '비만 2단계', '고도비만'],
    palette='Blues',
    ax=axes[0]
)
axes[0].set_title('인지장애 경험자의 BMI 분포')
axes[0].set_xlabel('BMI')
axes[0].set_ylabel('수')

# 라벨링
for p in plot1.patches:
    height = int(p.get_height())
    axes[0].annotate(f'{height}', (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='bottom', fontsize=10)

# 인지장애 미경험자
plot2 = sns.countplot(
    x=df_nco['BMI 수치'],
    order=['저체중', '정상', '과체중', '비만 1단계', '비만 2단계', '고도비만'],
    palette='Reds',
    ax=axes[1]
)
axes[1].set_title('인지장애 미경험자의 BMI 분포')
axes[1].set_xlabel('BMI')
axes[1].set_ylabel('수')

# 라벨링
for p in plot2.patches:
    height = int(p.get_height())
    axes[1].annotate(f'{height}', (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()

df_old.drop(columns=['BMI 수치'], inplace=True)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 고강도 운동 실천 빈도
def activity(x):
    if pd.isna(x) or x in [-1, 777, 888, 999]:
        return '삭제'
    elif x  == 0:
        return '전혀 안 함'
    elif 1 <= x <= 2:
        return '주 1~2회'
    elif 3 <= x <= 4:
        return '주 3~4회'
    elif 5 <= x <= 6:
        return '주 5~6회'
    else:
        return '매일'

df_old['고강도 신체활동 일'] = df_old['고강도 신체활동(일)'].apply(activity)

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# ─── 2. 데이터 분리 ───────────────────────────────────────
df_co  = df_old[df_old['인지장애 경험 여부'] == 1]
df_nco = df_old[df_old['인지장애 경험 여부'] == 2]

order = ['전혀 안 함', '주 1~2회', '주 3~4회', '주 5~6회', '매일']

# ─── 3. 기본 그래프 ───────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(6, 4), sharey=True)

# 경험자
counts_co = (
    df_co['고강도 신체활동 일'].value_counts(normalize=True)
    .reindex(order, fill_value=0) * 100
)
sns.barplot(x=counts_co.index, y=counts_co.values, palette='Blues', ax=axes[0])
axes[0].set_title('인지장애 경험자의 고강도 신체활동', fontsize=8)
axes[0].set_xlabel('빈도'); axes[0].set_ylabel('비율 (%)', fontsize=8)

# 미경험자
counts_nco = (
    df_nco['고강도 신체활동 일'].value_counts(normalize=True)
    .reindex(order, fill_value=0) * 100
)
sns.barplot(x=counts_nco.index, y=counts_nco.values, palette='Reds', ax=axes[1])
axes[1].set_title('인지장애 미경험자의 고강도 신체활동', fontsize=8)
axes[1].set_xlabel('빈도'); axes[1].set_ylabel('비율 (%)', fontsize=8)

# ─── 4. 막대 폭·라벨 재조정 ───────────────────────────────
for ax in axes:

    for p in ax.patches:
        new_w  = 0.4
        shift  = (p.get_width() - new_w) / 2
        p.set_width(new_w)
        p.set_x(p.get_x() + shift)
        ax.tick_params(axis='x', labelsize=7)
        h = p.get_height()
        ax.annotate(f'{h:.1f}%',
                    (p.get_x() + new_w / 2, h),
                    ha='center', va='bottom', fontsize=8)

plt.tight_layout()
plt.show()

# ─── 5. 임시 컬럼 제거 ───────────────────────────────────
df_old.drop(columns=['고강도 신체활동 일'], inplace=True)
```

### 고강도 신체활동 시간

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 한글, 마이너스 깨짐 방지~
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

df_old = df_old.copy()

# 운동실천 시간 나누기 
def activity_m(x):
    if pd.isna(x) or x in [-1, 777, 888, 999]:
        return '삭제'
    elif x < 20 :
        return '20분 미만'
    elif 20 <= x <= 39:
        return '20~39분'
    elif 40 <= x <= 59:
        return '40~59분'
    else:
        return '1시간 이상'

df_old['고강도 신체활동 분'] = df_old['고강도 신체활동(분)'].apply(activity_m)

# 데이터 분리
df_co = df_old[df_old['인지장애 경험 여부'] == 1]
df_nco = df_old[df_old['인지장애 경험 여부'] == 2]

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5), sharey=True)

# 인지장애 경험자
plot1 = sns.countplot(
    x=df_co['고강도 신체활동 분'],
    order=['20분 미만', '20~39분', '40~59분', '1시간 이상'],
    palette='Blues',
    ax=axes[0]
)
axes[0].set_title('인지장애 경험자의 고강도 신체활동 시간 분포')
axes[0].set_xlabel('고강도 신체활동 분')
axes[0].set_ylabel('수')

# 라벨링
for p in plot1.patches:
    height = int(p.get_height())
    axes[0].annotate(f'{height}', (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='bottom', fontsize=10)

# 인지장애 미경험자
plot2 = sns.countplot(
    x=df_nco['고강도 신체활동 분'],
    order=['20분 미만', '20~39분', '40~59분', '1시간 이상'],
    palette='Reds',
    ax=axes[1]
)
axes[1].set_title('인지장애 미경험자의 고강도 신체활동 시간 분포')
axes[1].set_xlabel('고강도 신체활동 분')
axes[1].set_ylabel('수')

# 라벨링
for p in plot2.patches:
    height = int(p.get_height())
    axes[1].annotate(f'{height}', (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()

df_old.drop(columns=['고강도 신체활동 분'], inplace= True)
```

```python
# 실 수면 시간
def sleep(x):
    if pd.isna(x) or x in [-1, 777, 888, 999]:
        return '삭제'
    elif x < 240:
        return '4시간 미만'
    elif 240 <= x <= 359:
        return '4~6시간'
    elif 360 <= x <= 479:
        return '6~8시간'
    elif 480 <= x <= 599:
        return '8~10시간'
    else:
        return '10시간 이상'

df_old['실 수면 시간 분'] = df_old['실수면(분)'].apply(sleep)

# 한글, 마이너스 깨짐 방지~
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 데이터 분리
df_co = df_old[df_old['인지장애 경험 여부'] == 1]
df_nco = df_old[df_old['인지장애 경험 여부'] == 2]

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5), sharey=True)

# 인지장애 경험자
plot1 = sns.countplot(
    x=df_co['실 수면 시간 분'],
    order=['4시간 미만', '4~6시간', '6~8시간', '8~10시간', '10시간 이상'],
    palette='Blues',
    ax=axes[0]
)
axes[0].set_title('인지장애 경험자의 실제 수면 시간')
axes[0].set_xlabel('실제 수면 시간')
axes[0].set_ylabel('수')

# 라벨링
for p in plot1.patches:
    height = int(p.get_height())
    axes[0].annotate(f'{height}', (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='bottom', fontsize=10)

# 인지장애 미경험자
plot2 = sns.countplot(
    x=df_nco['실 수면 시간 분'],
    order=['4시간 미만', '4~6시간', '6~8시간', '8~10시간', '10시간 이상'],
    palette='Reds',
    ax=axes[1]
)
axes[1].set_title('인지장애 미경험자의 실제 수면 시간')
axes[1].set_xlabel('실제 수면 시간 분')
axes[1].set_ylabel('수')

# 라벨링
for p in plot2.patches:
    height = int(p.get_height())
    axes[1].annotate(f'{height}', (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()

df_old.drop(columns=['실 수면 시간 분'], inplace=True)
```

```python
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 매핑용 라벨
label_수면유지 = {
    0.1 : '제외',
    1 : '1달간 없음',
    0.75 : '주 1회 미만',
    0.5 : '주 1~2회',
    0.25 : '주 3회 이상'
}

mask_valid = df_old['수면유지장애'].isin([1, 0.75, 0.5, 0.25])

df_co  = df_old[(df_old['인지장애 경험 여부'] == 1) & mask_valid]
df_nco = df_old[(df_old['인지장애 경험 여부'] == 2) & mask_valid]

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(6, 4), sharey=True)

counts_co = (
    df_co['수면유지장애'].value_counts(normalize=True) * 100
    ).rename(index=label_수면_통증).reindex(['1달간 없음', '주 1회 미만', '주 1~2회', '주 3회 이상'])
sns.barplot(x=counts_co.index, y=counts_co.values, palette='Blues', ax=axes[0])
axes[0].set_title('인지장애 경험자의 수면유지장애', fontsize=8)
axes[0].set_xlabel('응답'); axes[0].set_ylabel('비율 (%)', fontsize=8)

# 3‑2) 인지장애 미경험자
counts_nco = (
    df_nco['수면유지장애'].value_counts(normalize=True) * 100
    ).rename(index=label_수면_통증).reindex(['1달간 없음', '주 1회 미만', '주 1~2회', '주 3회 이상'])
sns.barplot(x=counts_nco.index, y=counts_nco.values, palette='Reds', ax=axes[1])
axes[1].set_title('인지장애 미경험자의 수면유지장애', fontsize=8)
axes[1].set_xlabel('응답'); axes[1].set_ylabel('비율 (%)', fontsize=8)

for ax in axes:                      
    for p in ax.patches:             
        new_w  = 0.4                 
        shift  = (p.get_width() - new_w) / 2
        p.set_width(new_w)
        p.set_x(p.get_x() + shift)
        h = p.get_height()      
        ax.tick_params(axis='x', labelsize=7)     
        ax.annotate(f'{h:.1f}%',
                    (p.get_x() + new_w/2, h),
                    ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.show()
```

```python
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 매핑용 라벨
label_수면_통증 = {
    0.1 : '제외',
    1 : '1달간 없음',
    0.75 : '주 1회 미만',
    0.5 : '주 1~2회',
    0.25 : '주 3회 이상'
}

mask_valid = df_old['수면 중 통증'].isin([1, 0.75, 0.5, 0.25])

df_co  = df_old[(df_old['인지장애 경험 여부'] == 1) & mask_valid]
df_nco = df_old[(df_old['인지장애 경험 여부'] == 2) & mask_valid]

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(6, 4), sharey=True)

counts_co = (
    df_co['수면 중 통증'].value_counts(normalize=True) * 100
    ).rename(index=label_수면_통증).reindex(['1달간 없음', '주 1회 미만', '주 1~2회', '주 3회 이상'])
sns.barplot(x=counts_co.index, y=counts_co.values, palette='Blues', ax=axes[0])
axes[0].set_title('인지장애 경험자의 수면 중 통증', fontsize=8)
axes[0].set_xlabel('응답'); axes[0].set_ylabel('비율 (%)', fontsize=8)

# 3‑2) 인지장애 미경험자
counts_nco = (
    df_nco['수면 중 통증'].value_counts(normalize=True) * 100
    ).rename(index=label_수면_통증).reindex(['1달간 없음', '주 1회 미만', '주 1~2회', '주 3회 이상'])
sns.barplot(x=counts_nco.index, y=counts_nco.values, palette='Reds', ax=axes[1])
axes[1].set_title('인지장애 미경험자의 수면 중 통증', fontsize=8)
axes[1].set_xlabel('응답'); axes[1].set_ylabel('비율 (%)', fontsize=8)

for ax in axes:                      
    for p in ax.patches:             
        new_w  = 0.4                 
        shift  = (p.get_width() - new_w) / 2
        p.set_width(new_w)
        p.set_x(p.get_x() + shift)
        h = p.get_height()      
        ax.tick_params(axis='x', labelsize=7)     
        ax.annotate(f'{h:.1f}%',
                    (p.get_x() + new_w/2, h),
                    ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.show()
```

```python
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 매핑용 라벨
label_수면질 = {
    0.1 : '제외',
    1 : '매우 높음',
    0.75 : '높음',
    0.5 : '낮음',
    0.25 : '매우 낮음'
}

mask_valid = df_old['수면 질'].isin([1, 0.75, 0.5, 0.25])

df_co  = df_old[(df_old['인지장애 경험 여부'] == 1) & mask_valid]
df_nco = df_old[(df_old['인지장애 경험 여부'] == 2) & mask_valid]

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(6, 4), sharey=True)

counts_co = (
    df_co['수면 질'].value_counts(normalize=True) * 100
    ).rename(index=label_수면질).reindex(['매우 높음', '높음', '낮음', '매우 낮음'])
sns.barplot(x=counts_co.index, y=counts_co.values, palette='Blues', ax=axes[0])
axes[0].set_title('인지장애 경험자의 수면 질', fontsize=8)
axes[0].set_xlabel('응답'); axes[0].set_ylabel('비율 (%)', fontsize=8)

# 3‑2) 인지장애 미경험자
counts_nco = (
    df_nco['수면 질'].value_counts(normalize=True) * 100
    ).rename(index=label_수면질).reindex(['매우 높음', '높음', '낮음', '매우 낮음'])
sns.barplot(x=counts_nco.index, y=counts_nco.values, palette='Reds', ax=axes[1])
axes[1].set_title('인지장애 미경험자의 수면 질', fontsize=8)
axes[1].set_xlabel('응답'); axes[1].set_ylabel('비율 (%)', fontsize=8)

for ax in axes:                      
    for p in ax.patches:             
        new_w  = 0.4                 
        shift  = (p.get_width() - new_w) / 2
        p.set_width(new_w)
        p.set_x(p.get_x() + shift)

        h = p.get_height()           
        ax.annotate(f'{h:.1f}%',
                    (p.get_x() + new_w/2, h),
                    ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.show()
```

```python
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 매핑용 라벨
label_우울감_경험 = {
    1 : '예',
    2 : '아니오',
    -1 : '삭제'
}

mask_valid = df_old['우울감 경험'].isin([1, 2])

df_co  = df_old[(df_old['인지장애 경험 여부'] == 1) & mask_valid]
df_nco = df_old[(df_old['인지장애 경험 여부'] == 2) & mask_valid]

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(6, 4), sharey=True)

counts_co = (
    df_co['우울감 경험'].value_counts(normalize=True) * 100
    ).rename(index=label_우울감_경험).reindex(['아니오', '예'])
sns.barplot(x=counts_co.index, y=counts_co.values, palette='Blues', ax=axes[0])
axes[0].set_title('인지장애 경험자의 우울감 경험 여부', fontsize=8)
axes[0].set_xlabel('응답'); axes[0].set_ylabel('비율 (%)', fontsize=8)

# 3‑2) 인지장애 미경험자
counts_nco = (
    df_nco['우울감 경험'].value_counts(normalize=True) * 100
    ).rename(index=label_우울감_경험).reindex(['아니오', '예'])
sns.barplot(x=counts_nco.index, y=counts_nco.values, palette='Reds', ax=axes[1])
axes[1].set_title('인지장애 미경험자의 우울감 경험 여부', fontsize=8)
axes[1].set_xlabel('응답'); axes[1].set_ylabel('비율 (%)', fontsize=8)

for ax in axes:                      
    for p in ax.patches:             
        new_w  = 0.4                 
        shift  = (p.get_width() - new_w) / 2
        p.set_width(new_w)
        p.set_x(p.get_x() + shift)

        h = p.get_height()           
        ax.annotate(f'{h:.1f}%',
                    (p.get_x() + new_w/2, h),
                    ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

label_우울감_자살 = {1: '아니오', 0.75: '여러 날', 0.5: '일주일 이상',
                  0.25: '매일', 0.1: '삭제'}

# 유효 값(1·0.75·0.5·0.25)만 남기기  ← 여기!
mask_valid = df_old['자살충동'].isin([1, 0.75, 0.5, 0.25])

df_co  = df_old[(df_old['인지장애 경험 여부'] == 1) & mask_valid]
df_nco = df_old[(df_old['인지장애 경험 여부'] == 2) & mask_valid]

fig, axes = plt.subplots(1, 2, figsize=(6, 4), sharey=True)

# ── 경험자 ───────────────────────────────────────────
counts_co = (
    df_co['자살충동'].value_counts(normalize=True)*100
).rename(index=label_우울감_자살) \
 .reindex(['아니오','여러 날','일주일 이상','매일'], fill_value=0)   # ← 여기!
sns.barplot(x=counts_co.index, y=counts_co.values, palette='Blues', ax=axes[0])
axes[0].set_title('인지장애 경험자의 자살충동 빈도', fontsize=8)
axes[0].set_xlabel('응답'); axes[0].set_ylabel('비율 (%)', fontsize=8)

# ── 미경험자 ─────────────────────────────────────────
counts_nco = (
    df_nco['자살충동'].value_counts(normalize=True)*100
).rename(index=label_우울감_자살) \
 .reindex(['아니오','여러 날','일주일 이상','매일'], fill_value=0)
sns.barplot(x=counts_nco.index, y=counts_nco.values, palette='Reds', ax=axes[1])
axes[1].set_title('인지장애 미경험자의 자살충동 빈도', fontsize=8)
axes[1].set_xlabel('응답'); axes[1].set_ylabel('비율 (%)', fontsize=8)

# 막대 폭·라벨 재조정
for ax in axes:
    for p in ax.patches:
        new_w  = 0.4
        shift  = (p.get_width() - new_w)/2
        p.set_width(new_w); p.set_x(p.get_x()+shift)

        h = p.get_height()
        ax.annotate(f'{h:.1f}%', (p.get_x()+new_w/2, h),
                    ha='center', va='bottom', fontsize=8)

plt.tight_layout()
plt.show()
```

### 우울증상 위험도

```python
df_old['우울증상'].value_counts()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 우울증상 빈도
def melancholy(x):
    if pd.isna(x) or x in [-5]:
        return '삭제'
    elif x  == 5:
        return '매우 안전'
    elif 6 <= x <= 10:
        return '보통'
    elif 11 <= x <= 16:
        return '위험'
    else:
        return '매우 위험'

df_old['우울증상 점수'] = df_old['우울증상'].apply(melancholy)

# 한글, 마이너스 깨짐 방지~
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 데이터 분리
df_co = df_old[df_old['인지장애 경험 여부'] == 1]
df_nco = df_old[df_old['인지장애 경험 여부'] == 2]

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5), sharey=True)

# 인지장애 경험자
plot1 = sns.countplot(
    x=df_co['우울증상 점수'],
    order=['매우 안전', '보통', '위험', '매우 위험'],
    palette='Blues',
    ax=axes[0]
)
axes[0].set_title('인지장애 경험자의 우울증상 점수 분포')
axes[0].set_xlabel('우울점수')
axes[0].set_ylabel('수')

# 라벨링
for p in plot1.patches:
    height = int(p.get_height())
    axes[0].annotate(f'{height}', (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='bottom', fontsize=10)

# 인지장애 미경험자
plot2 = sns.countplot(
    x=df_nco['우울증상 점수'],
    order=['매우 안전', '보통', '위험', '매우 위험'],
    palette='Reds',
    ax=axes[1]
)
axes[1].set_title('인지장애 미경험자의 우울증상 점수 분포')
axes[1].set_xlabel('우울점수')
axes[1].set_ylabel('수')

# 라벨링
for p in plot2.patches:
    height = int(p.get_height())
    axes[1].annotate(f'{height}', (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()

df_old.drop(columns=['우울증상 점수'], inplace=True)
```

#### 비율

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ─── 1. 전처리 ────────────────────────────────────────────
def melancholy(x):
    if pd.isna(x) or x in [-5]:
        return '삭제'
    elif x == 5:
        return '매우 안전'
    elif 6 <= x <= 10:
        return '보통'
    elif 11 <= x <= 16:
        return '위험'
    else:
        return '매우 위험'

df_old['우울증상 점수'] = df_old['우울증상'].apply(melancholy)

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# ─── 2. 데이터 분리 ───────────────────────────────────────
df_co  = df_old[df_old['인지장애 경험 여부'] == 1]
df_nco = df_old[df_old['인지장애 경험 여부'] == 2]

order = ['매우 안전', '보통', '위험', '매우 위험']

# ─── 3. 기본 그래프 ───────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(6, 4), sharey=True)

# 경험자
counts_co = (
    df_co['우울증상 점수'].value_counts(normalize=True)
    .reindex(order, fill_value=0) * 100
)
sns.barplot(x=counts_co.index, y=counts_co.values, palette='Blues', ax=axes[0])
axes[0].set_title('인지장애 경험자의 우울증상 점수 분포', fontsize=10)
axes[0].set_xlabel('우울증상 점수'); axes[0].set_ylabel('비율 (%)', fontsize=8)

# 미경험자
counts_nco = (
    df_nco['우울증상 점수'].value_counts(normalize=True)
    .reindex(order, fill_value=0) * 100
)
sns.barplot(x=counts_nco.index, y=counts_nco.values, palette='Reds', ax=axes[1])
axes[1].set_title('인지장애 미경험자의 우울증상 점수 분포', fontsize=10)
axes[1].set_xlabel('우울증상 점수'); axes[1].set_ylabel('비율 (%)', fontsize=8)

# ─── 4. 막대 폭·라벨 재조정 ───────────────────────────────
for ax in axes:

    for p in ax.patches:
        new_w  = 0.4
        shift  = (p.get_width() - new_w) / 2
        p.set_width(new_w)
        p.set_x(p.get_x() + shift)

        h = p.get_height()
        ax.annotate(f'{h:.1f}%',
                    (p.get_x() + new_w / 2, h),
                    ha='center', va='bottom', fontsize=8)

plt.tight_layout()
plt.show()

# ─── 5. 임시 컬럼 제거 ───────────────────────────────────
df_old.drop(columns=['우울증상 점수'], inplace=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

label_당뇨_진단 = {1: '예', 2: '아니오', -1: '삭제'}

mask_valid = df_old['고혈압 진단'].isin([1, 2])

df_co  = df_old[(df_old['인지장애 경험 여부'] == 1) & mask_valid]
df_nco = df_old[(df_old['인지장애 경험 여부'] == 2) & mask_valid]

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(6, 4), sharey=True)

counts_co = (
    df_co['고혈압 진단'].value_counts(normalize=True) * 100
    ).rename(index=label_당뇨_진단).reindex(['아니오', '예'])
sns.barplot(x=counts_co.index, y=counts_co.values, palette='Blues', ax=axes[0])
axes[0].set_title('인지장애 경험자의 고혈압 진단 여부', fontsize=8)
axes[0].set_xlabel('응답'); axes[0].set_ylabel('비율 (%)', fontsize=8)

# 3‑2) 인지장애 미경험자
counts_nco = (
    df_nco['고혈압 진단'].value_counts(normalize=True) * 100
    ).rename(index=label_당뇨_진단).reindex(['아니오', '예'])
sns.barplot(x=counts_nco.index, y=counts_nco.values, palette='Reds', ax=axes[1])
axes[1].set_title('인지장애 미경험자의 고혈압 진단 여부', fontsize=8)
axes[1].set_xlabel('응답'); axes[1].set_ylabel('비율 (%)', fontsize=8)

for ax in axes:                      
    for p in ax.patches:             
        new_w  = 0.4                 
        shift  = (p.get_width() - new_w) / 2
        p.set_width(new_w)
        p.set_x(p.get_x() + shift)

        h = p.get_height()           
        ax.annotate(f'{h:.1f}%',
                    (p.get_x() + new_w/2, h),
                    ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.show()
```

### 당뇨 진단 여부

```python
df_old['당뇨병 진단'].describe()
```

#### 비율

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# ──────────────────── 1. 라벨 사전 ────────────────────
label_당뇨_진단 = {1: '예', 2: '아니오', -1: '삭제'}

# ──────────────────── 2. 유효 응답(1·2)만 추출 ────────────────────
mask_valid = df_old['당뇨병 진단'].isin([1, 2])

df_co  = df_old[(df_old['인지장애 경험 여부'] == 1) & mask_valid]
df_nco = df_old[(df_old['인지장애 경험 여부'] == 2) & mask_valid]

# ──────────────────── 3. 기본 그래프 ────────────────────
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(6, 4), sharey=True)

# 3‑1) 인지장애 경험자
counts_co = (
    df_co['당뇨병 진단'].value_counts(normalize=True) * 100
    ).rename(index=label_당뇨_진단).reindex(['아니오', '예'])
sns.barplot(x=counts_co.index, y=counts_co.values, palette='Blues', ax=axes[0])
axes[0].set_title('인지장애 경험자의 당뇨병 진단 여부', fontsize=8)
axes[0].set_xlabel('응답'); axes[0].set_ylabel('비율 (%)', fontsize=8)

# 3‑2) 인지장애 미경험자
counts_nco = (
    df_nco['당뇨병 진단'].value_counts(normalize=True) * 100
    ).rename(index=label_당뇨_진단).reindex(['아니오', '예'])
sns.barplot(x=counts_nco.index, y=counts_nco.values, palette='Reds', ax=axes[1])
axes[1].set_title('인지장애 미경험자의 당뇨병 진단 여부', fontsize=8)
axes[1].set_xlabel('응답'); axes[1].set_ylabel('비율 (%)', fontsize=8)

for ax in axes:                      
    for p in ax.patches:             
        new_w  = 0.4                 
        shift  = (p.get_width() - new_w) / 2
        p.set_width(new_w)
        p.set_x(p.get_x() + shift)

        h = p.get_height()           
        ax.annotate(f'{h:.1f}%',
                    (p.get_x() + new_w/2, h),
                    ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.show()
```

```python
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 매핑용 라벨
label_당화혈색소 = {
    1 : '당화혈색소를 모름',
    2 : '4회 이상',
    3 : '3회',
    4 : '2회',
    5 : '1회',
    6 : '측정 안함',
    -1 : '삭제'
}

# 인지장애 경험자, 미경험자 나누기
df_co = df_old[df_old['인지장애 경험 여부'] == 1]  # 인지장애 경험자
df_nco = df_old[df_old['인지장애 경험 여부'] == 2]  # 인지장애 미경험자


fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5), sharey=True)

# 인지장애 경험자
plot1= sns.countplot(
    x=df_co['당화혈색소 검사'].map(label_당화혈색소),
    order=['1회', '2회', '3회', '4회 이상', '당화혈색소를 모름'],
    palette='Blues',
    ax=axes[0]
)
axes[0].set_title('인지장애 경험자의 당화혈색소 검사 여부')
axes[0].set_xlabel('응답')
axes[0].set_ylabel('수')

# 인지장애 경험자 데이터 라벨링 추가
for p in plot1.patches:
    height = int(p.get_height())
    axes[0].annotate(f'{height}', (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='bottom', fontsize=10)

# 인지장애 미경험자
plot2 = sns.countplot(
    x=df_nco['당화혈색소 검사'].map(label_당화혈색소),
    order=['1회', '2회', '3회', '4회 이상', '당화혈색소를 모름'],
    palette='Reds',
    ax=axes[1]
)
axes[1].set_title('인지장애 미경험자의 당화혈색소 검사 여부')
axes[1].set_xlabel('응답')
axes[1].set_ylabel('수')

# 인지장애 미경험자 데이터 라벨링 추가
for p in plot2.patches:
    height = int(p.get_height())
    axes[1].annotate(f'{height}', (p.get_x() + p.get_width() / 2., height),
                     ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

#### 비율

```python
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 매핑용 라벨
label_당화혈색소 = {
    1 : '당화혈색소를 모름',
    2 : '4회 이상',
    3 : '3회',
    4 : '2회',
    5 : '1회',
    6 : '측정 안함',
    -1 : '삭제'
}

# 인지장애 경험자, 미경험자 나누기
df_co = df_old[df_old['인지장애 경험 여부'] == 1]
df_nco = df_old[df_old['인지장애 경험 여부'] == 2]

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5), sharey=True)

# 인지장애 경험자 비율 계산 및 시각화
counts_co = df_co.loc[df_co['당화혈색소 검사'] != -1,
                      '당화혈색소 검사'] \
                 .value_counts(normalize=True) * 100
mapped_co = counts_co.rename(index=label_당화혈색소).reindex(['1회', '2회', '3회', '4회 이상', '당화혈색소를 모름'])

sns.barplot(x=mapped_co.index, y=mapped_co.values, palette='Blues', ax=axes[0])
axes[0].set_title('인지장애 경험자의 당화혈색소 검사 여부')
axes[0].set_xlabel('응답')
axes[0].set_ylabel('비율')

for i, v in enumerate(mapped_co.values):
    axes[0].annotate(f'{v:.1f}%', (i, v), ha='center', va='bottom', fontsize=10)

# 인지장애 미경험자 비율 계산 및 시각화
counts_nco = df_nco.loc[df_nco['당화혈색소 검사'] != -1,
                      '당화혈색소 검사'] \
                 .value_counts(normalize=True) * 100
mapped_nco = counts_nco.rename(index=label_당화혈색소).reindex(['1회', '2회', '3회', '4회 이상', '당화혈색소를 모름'])

sns.barplot(x=mapped_nco.index, y=mapped_nco.values, palette='Reds', ax=axes[1])
axes[1].set_title('인지장애 미경험자의 당화혈색소 검사 여부')
axes[1].set_xlabel('응답')
axes[1].set_ylabel('비율')

for i, v in enumerate(mapped_nco.values):
    axes[1].annotate(f'{v:.1f}%', (i, v), ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```

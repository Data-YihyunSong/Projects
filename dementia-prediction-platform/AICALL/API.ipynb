```python
# .wav 파일 재생하는 함수
from pydub import AudioSegment
import simpleaudio as sa

def play_audio(file_path):
    audio = AudioSegment.from_file(file_path, format='wav')
    play_obj = sa.play_buffer(
        audio.raw_data,
        num_channels=audio.channels,
        bytes_per_sample=audio.sample_width,
        sample_rate=audio.frame_rate
    )
    play_obj.wait_done()
```

```python
# 사용자의 응답을 감지하여 응답이 끝나면 녹음 자동 종료 함수
import sounddevice as sd
import webrtcvad
import collections
import wave
from scipy.io.wavfile import write

def record_until_silence(filename='output.wav', sample_rate=16000, silence_limit=2.0, aggressiveness=2):
    vad = webrtcvad.Vad(aggressiveness)
    frame_duration = 30  # ms
    frame_size = int(sample_rate * frame_duration / 1000)
    buffer = collections.deque(maxlen=int(silence_limit * 1000 / frame_duration))
    audio_frames = []

    def callback(indata, frames, time, status):
        nonlocal audio_frames, buffer, recording_done
        pcm_data = indata[:, 0].tobytes()
        is_speech = vad.is_speech(pcm_data, sample_rate)
        buffer.append(is_speech)
        audio_frames.append(pcm_data)
        if len(buffer) == buffer.maxlen and not any(buffer):
            recording_done = True
            raise sd.CallbackStop()

    print('녹음 시작 (음성 감지)')
    recording_done = False
    try:
        with sd.InputStream(samplerate=sample_rate, channels=1, dtype='int16', callback=callback,
                            blocksize=frame_size):
            while not recording_done:
                sd.sleep(100)
    except sd.CallbackStop:
        pass

    wf = wave.open(filename, 'wb')
    wf.setnchannels(1)
    wf.setsampwidth(2)
    wf.setframerate(sample_rate)
    wf.writeframes(b''.join(audio_frames))
    wf.close()
    print(f'녹음 완료: {filename}')
```

```python
# NAVER CLOVA STT API 활용하여 사용자의 음성 응답을 텍스트로 변환
import requests
import json
import uuid # 고유 식별자 생성, 데이터베이스 키나 객체 식별할 때 사용
import os # 운영체제에서 사용할 수 있는 기능을 파이썬에서 사용하게 해주는 기본 모듈임

def clova_stt(audio_path):
    invoke_url = 'https://clovaspeech-gw.ncloud.com/external/v1/11850/592d8cec2aab7ab960eb325ec6b195666103d8704d55939edc9f677b94d0c0e1/recognizer/upload'
    secret_key = os.getenv('NCLOUD_SECRET_KEY')

    
    request_body = {
        'language': 'ko-KR',
        'completion': 'sync',
        'fullText': True,
        'diarization' : {'enable' : False}
    }

    headers = {
        'Accept': 'application/json;UTF-8',
        'X-CLOVASPEECH-API-KEY': secret_key
    }

    files = {
    'media': ('filename.wav', 
              open(audio_path, 'rb'), 
              'application/octet-stream'),
    'params': (None, 
               json.dumps(request_body, ensure_ascii=False).encode('UTF-8'), 
               'application/json')
}

    response = requests.post(headers=headers, url=invoke_url, files=files)

    print('상태코드:', response.status_code)
    print('응답 내용:', response.text)

    try:
        result = response.json()
    except Exception as e:
        return '[STT 실패] 응답 파싱 오류'

    return result.get('text', '[STT 실패] text 키 없음')
```

```python
# 각 문항별 점수로 채점하는 방식
from openai import OpenAI

client = OpenAI()

def recog_answer(question_text, user_text, expected_answer, score_info):
    prompt = f'''
    질문: {question_text}
    사용자의 응답: {user_text}
    정답 기준: {expected_answer}
    점수: {score_info}

    판단 기준:
    - '사용자의 응답'이 '정답 기준'과 하나 이상 의미상 동일하거나 그 단어를 포함하고 있을 경우 '점수'를 그대로 반환해줘.
    - 형태적인 변형(예: 띄어쓰기, 조사 생략, 단어 순서 변화, 숫자 표현의 한글/숫자 등)은 모두 허용해.
    - 점수는 정수(int)만 출력해줘. (예: 8)
    - '정답', '오답' 등의 단어는 절대 출력하지 마.
    '''

    response = client.chat.completions.create(
        model='gpt-4o',
        messages=[{'role': 'user', 'content': prompt}],
        temperature=0.0
    )
    result = response.choices[0].message.content.strip()
    return result
```

```python
from datetime import datetime
import re

# 사용자 정보 입력
name = input('이름을 입력하세요: ')
birth_str = input('생년월일을 입력해주세요 (예: 19900101): ')
birth = datetime.strptime(birth_str, '%Y%m%d')
birth_str_fmt = birth.strftime('%Y%m%d')

# 초기 user 데이터프레임 생성
user = pd.DataFrame([{
    '이름': name,
    '생년월일': birth
}])

# 응답 및 결과 저장용 dict
responses = {}
results = {}

total_score = 0

skip_q6_subquestions = False

def parse_score(score_str):
    try:
        m = re.search(r'\d+', score_str)
        if m:
            return int(m.group())
        else:
            return 0
    except Exception as e:
        print(f'[경고] 점수 파싱 실패: {score_str} ({e})')
        return 0

# CIST 검사
for item in script:
    label = item['label']
    file_path = item['file']

    # Q6 하위 문항 skip
    if label.startswith('Q6_') and skip_q6_subquestions:
        print(f'[{label}] Q6 정답이므로 건너뜀')
        responses[label] = ''
        results[label + "_점수"] = 0
        continue

    # 오디오 재생
    play_audio(file_path)

    if 'answer' in item:
        rec_path = f"responses/{name}_{birth_str_fmt}_{label}_user.wav"
        if label == 'Q7':
            record_Q7(rec_path, duration=60)
        else:
            record_until_silence(rec_path)

        user_text = clova_stt(rec_path)

        # 점수 산출
        score_info = item.get('score', 0)
        
        # Q3, Q3_1 점수 없음
        if score_info == 0 or score_info is None:
            print(f'[{label}] 채점 없음: 응답만 저장')
            responses[label] = user_text
            results[label + '_점수'] = 0
            continue

        score_info_str = str(score_info)

        raw_score = recog_answer(
            question_text=item.get('content', ''),
            user_text=user_text,
            expected_answer=item.get('answer', ''),
            score_info=score_info_str
        )

        int_score = parse_score(raw_score)
        print(f'[{label}] 응답: {user_text}, GPT 점수: {raw_score}, 파싱 점수: {int_score}')

        responses[label] = user_text
        results[label + '_점수'] = int_score
        
        total_score += int_score

        if label == 'Q6':
            skip_q6_subquestions = (int_score > 0)

# user.csv에 점수와 응답 추가
for label in responses:
    user[label] = responses[label]
    user[label + '_점수'] = results.get(label + '_점수', 0)

# 최종 점수 컬럼 추가
user['총 획득 점수'] = total_score
```

```python
import pandas as pd
import numpy as np

df_174 = pd.read_csv('174_수정_1.csv')

# 교육 수준 → 교육년수 매핑
edu_years = {
    1: 0,   # 무학
    3: 6,   # 초
    4: 9,   # 중
    5: 12,  # 고
    6: 16,  # 대
    7: 16,
    8: 16
}

# 졸업 여부 1,2 → 현재 수준, 3,4 → 직전 수준
def get_edu_year(row):
    level = row['교육 수준']
    grad = row['졸업 여부']

    # 교육 수준이 무학일 경우는 무조건 0년으로 간주
    if level == 1:
        return 0

    if pd.isna(level) or pd.isna(grad):
        return np.nan

    if grad in [1, 2]:  # 현재 수준 졸업
        return edu_years.get(level, np.nan)
    
    elif grad in [3, 4]:  # 직전 수준 졸업
        prev_levels = sorted([k for k in edu_years if k < level], reverse=True)
        if prev_levels:
            prev_level = prev_levels[0]
            return edu_years.get(prev_level, np.nan)
        return np.nan

    else:
        return edu_years.get(level, np.nan)

# 파생변수 추가
df_174['교육년수'] = df_174.apply(get_edu_year, axis=1)

df_174.head()
```
